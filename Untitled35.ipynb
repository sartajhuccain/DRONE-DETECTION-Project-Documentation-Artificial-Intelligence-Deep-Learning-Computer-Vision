```python
import json

coco_candidates = []

for json_path in json_files:
    try:
        with open(json_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        if (
            isinstance(data, dict)
            and "images" in data
            and "annotations" in data
            and "categories" in data
        ):
            coco_candidates.append((json_path, data))

            print("\nCOCO annotation found:", json_path)
            print("Images:", len(data["images"]))
            print("Annotations:", len(data["annotations"]))
            print("Categories:", data["categories"])

    except (json.JSONDecodeError, UnicodeDecodeError):
        pass

print("\nTotal COCO annotation files:", len(coco_candidates))
```

```python
import pandas as pd
from PIL import Image

csv_rows = []

if not coco_candidates:
    print("COCO annotations nahi milin.")
else:
    for json_path, data in coco_candidates:
        image_info = {
            item["id"]: item
            for item in data["images"]
        }

        categories = {
            item["id"]: item["name"]
            for item in data["categories"]
        }

        # Image IDs ke saath annotations connect karein
        for ann in data["annotations"]:
            if ann.get("iscrowd", 0) == 1:
                continue

            image_id = ann.get("image_id")
            info = image_info.get(image_id)

            if info is None:
                continue

            bbox = ann.get("bbox", [])

            if len(bbox) != 4:
                continue

            x, y, w, h = map(float, bbox)

            if w <= 0 or h <= 0:
                continue

            category_id = ann.get("category_id")

            csv_rows.append({
                "annotation_file": str(json_path),
                "image_id": image_id,
                "image_file": info.get("file_name", ""),
                "width": info.get("width"),
                "height": info.get("height"),
                "class_id": category_id,
                "class_name": categories.get(
                    category_id, "unknown"
                ),
                "x_min": x,
                "y_min": y,
                "x_max": x + w,
                "y_max": y + h,
                "bbox_width": w,
                "bbox_height": h
            })

    annotations_df = pd.DataFrame(csv_rows)

    if not annotations_df.empty:
        CSV_PATH = "/content/drone_annotations.csv"

        annotations_df.to_csv(CSV_PATH, index=False)

        print("CSV successfully created!")
        print("CSV path:", CSV_PATH)
        print("Total annotation rows:", len(annotations_df))

        display(annotations_df.head(10))
    else:
        print("Koi valid bounding-box annotation nahi mili.")
```

```python
# ============================================
# DRONE DETECTION - FIX DATASET PATHS
# ============================================

import os
import json
import shutil
import random
import yaml
from pathlib import Path
from collections import defaultdict
from PIL import Image

random.seed(42)

# Dataset path
DATASET_DIR = Path(path)

print("Dataset directory:", DATASET_DIR)
print("Directory exists:", DATASET_DIR.exists())

if not DATASET_DIR.exists():
    raise FileNotFoundError(
        "Dataset directory nahi mila. Kaggle dataset dobara download karein."
    )

# Image extensions
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

# Find all images
image_files = [
    p for p in DATASET_DIR.rglob("*")
    if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
]

# Find COCO JSON annotation files
json_files = [
    p for p in DATASET_DIR.rglob("*.json")
    if p.is_file()
]

print("\nTotal images found:", len(image_files))
print("JSON files found:", len(json_files))

for p in json_files:
    print("JSON:", p.relative_to(DATASET_DIR))

if not image_files:
    raise ValueError(
        "Dataset mein images nahi milin. Dataset file listing check karein."
    )

# Load COCO annotations
coco_data = []

for json_path in json_files:
    try:
        with open(json_path, "r", encoding="utf-8") as f:
            data = json.load(f)

        if all(
            key in data
            for key in ["images", "annotations", "categories"]
        ):
            coco_data.append((json_path, data))

    except Exception as e:
        print("JSON skip:", json_path.name, str(e))

print("\nCOCO annotation files:", len(coco_data))

if not coco_data:
    raise ValueError(
        "COCO annotations nahi milin. Is dataset ka actual format "
        "check karna zaroori hai. Training abhi start nahi ki ja sakti."
    )

# Prepare output directories
OUTPUT_DIR = Path("/content/drone_yolo")

for split in ["train", "val", "test"]:
    (OUTPUT_DIR / "images" / split).mkdir(
        parents=True, exist_ok=True
    )
    (OUTPUT_DIR / "labels" / split).mkdir(
        parents=True, exist_ok=True
    )

# Remove files from any previous conversion
for split in ["train", "val", "test"]:
    for folder in [
        OUTPUT_DIR / "images" / split,
        OUTPUT_DIR / "labels" / split
    ]:
        for f in folder.iterdir():
            if f.is_file():
                f.unlink()

# Build image lookup by basename
image_lookup = defaultdict(list)

for img in image_files:
    image_lookup[img.name].append(img)

def find_image(file_name):
    candidate = Path(str(file_name))

    # Search relative to dataset directory
    direct = DATASET_DIR / candidate

    if direct.is_file():
        return direct

    # Search by basename
    matches = image_lookup.get(candidate.name, [])

    if matches:
        return matches[0]

    return None

# Read categories
categories = {}

for _, data in coco_data:
    for cat in data["categories"]:
        categories[cat["id"]] = cat["name"]

if not categories:
    raise ValueError("COCO categories empty hain.")

category_ids = sorted(categories.keys())

category_map = {
    old_id: new_id
    for new_id, old_id in enumerate(category_ids)
}

class_names = [
    categories[cid] for cid in category_ids
]

print("\nClasses:", class_names)

# Collect all image records and annotations
records = []

for json_path, data in coco_data:

    annotations_by_image = defaultdict(list)

    for ann in data["annotations"]:
        if ann.get("iscrowd", 0) == 1:
            continue

        bbox = ann.get("bbox", [])

        if len(bbox) != 4:
            continue

        x, y, w, h = map(float, bbox)

        if w <= 0 or h <= 0:
            continue

        category_id = ann.get("category_id")

        if category_id not in category_map:
            continue

        annotations_by_image[ann["image_id"]].append(
            (category_map[category_id], x, y, w, h)
        )

    for info in data["images"]:
        records.append({
            "json_path": json_path,
            "info": info,
            "annotations": annotations_by_image.get(
                info["id"], []
            )
        })

# Remove duplicate image records
unique_records = {}
for record in records:
    key = (
        str(record["json_path"]),
        record["info"]["id"]
    )
    unique_records[key] = record

records = list(unique_records.values())

print("\nTotal annotated image records:", len(records))

# Split dataset randomly if there is no reliable predefined split
random.shuffle(records)

n = len(records)

if n < 3:
    raise ValueError(
        f"Sirf {n} annotated images mili hain. "
        "Train/validation/test split ke liye dataset check karein."
    )

n_train = max(1, int(n * 0.70))
n_val = max(1, int(n * 0.20))

if n_train + n_val >= n:
    n_train = n - 2
    n_val = 1

split_records = {
    "train": records[:n_train],
    "val": records[n_train:n_train + n_val],
    "test": records[n_train + n_val:]
}

used_images = {
    "train": 0,
    "val": 0,
    "test": 0
}

missing_images = 0

# Copy images and generate YOLO labels
for split, split_items in split_records.items():

    for idx, record in enumerate(split_items):

        info = record["info"]
        img_path = find_image(info["file_name"])

        if img_path is None:
            missing_images += 1
            continue

        try:
            with Image.open(img_path) as im:
                width, height = im.size
        except Exception:
            missing_images += 1
            continue

        if width <= 0 or height <= 0:
            missing_images += 1
            continue

        unique_name = f"{split}_{idx:06d}"

        destination_image = (
            OUTPUT_DIR / "images" / split /
            f"{unique_name}{img_path.suffix.lower()}"
        )

        destination_label = (
            OUTPUT_DIR / "labels" / split /
            f"{unique_name}.txt"
        )

        shutil.copy2(img_path, destination_image)

        label_lines = []

        for class_id, x, y, w, h in record["annotations"]:

            x1 = max(0.0, min(x, width))
            y1 = max(0.0, min(y, height))
            x2 = max(0.0, min(x + w, width))
            y2 = max(0.0, min(y + h, height))

            bw = x2 - x1
            bh = y2 - y1

            if bw <= 0 or bh <= 0:
                continue

            xc = ((x1 + x2) / 2) / width
            yc = ((y1 + y2) / 2) / height

            nw = bw / width
            nh = bh / height

            label_lines.append(
                f"{class_id} {xc:.6f} {yc:.6f} "
                f"{nw:.6f} {nh:.6f}"
            )

        destination_label.write_text(
            "\n".join(label_lines),
            encoding="utf-8"
        )

        used_images[split] += 1

print("\n========== DATASET SUMMARY ==========")

for split in ["train", "val", "test"]:
    print(
        split.upper(),
        "Images:", used_images[split],
        "Label files:", len(
            list((OUTPUT_DIR / "labels" / split).glob("*.txt"))
        )
    )

print("Missing images:", missing_images)
print("Output directory:", OUTPUT_DIR)

if used_images["train"] == 0 or used_images["val"] == 0:
    raise ValueError(
        "Training ya validation images zero hain. "
        "Image paths aur annotations check karein."
    )

# Create YOLO YAML
yaml_data = {
    "path": str(OUTPUT_DIR),
    "train": "images/train",
    "val": "images/val",
    "names": {
        i: name for i, name in enumerate(class_names)
    }
}

if used_images["test"] > 0:
    yaml_data["test"] = "images/test"

YOLO_YAML_PATH = OUTPUT_DIR / "drone.yaml"

with open(YOLO_YAML_PATH, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        yaml_data,
        f,
        sort_keys=False
    )

print("\nYOLO YAML created:", YOLO_YAML_PATH)
print("Dataset preparation completed!")
```

```python
import yaml

yaml_data = {
    "path": str(OUTPUT_DIR),
    "train": "images/train",
    "val": "images/val",
    "names": {
        i: name for i, name in enumerate(class_names)
    }
}

if used_images["test"] > 0:
    yaml_data["test"] = "images/test"

YOLO_YAML_PATH = OUTPUT_DIR / "drone.yaml"

with open(YOLO_YAML_PATH, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        yaml_data,
        f,
        sort_keys=False,
        allow_unicode=True
    )

print("Configuration file:", YOLO_YAML_PATH)

with open(YOLO_YAML_PATH, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
from ultralytics import YOLO
import torch

device = 0 if torch.cuda.is_available() else "cpu"

model = YOLO("yolo11n.pt")

print("Model loaded successfully!")
print("Training device:", device)
```

```python
EPOCHS = 30
IMAGE_SIZE = 640
BATCH_SIZE = 8

results = model.train(
    data=str(YOLO_YAML_PATH),
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    device=device,
    workers=2,
    patience=8,
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    cos_lr=True,
    seed=42,
    project="/content/drone_runs",
    name="drone_detection",
    exist_ok=True,
    plots=True,
    save=True,
    verbose=True
)

print("Training complete!")
```

```python
from pathlib import Path
from ultralytics import YOLO

BEST_MODEL_PATH = Path(
    "/content/drone_runs/drone_detection/weights/best.pt"
)

if not BEST_MODEL_PATH.exists():
    raise FileNotFoundError(
        f"Trained model nahi mila: {BEST_MODEL_PATH}\n"
        "Cell 13 ki training complete hui thi ya nahi check karein."
    )

best_model = YOLO(str(BEST_MODEL_PATH))

print("Best model loaded:", BEST_MODEL_PATH)
```

```python
metrics = best_model.val(
    data=str(YOLO_YAML_PATH),
    split="val",
    imgsz=640,
    device=device,
    plots=True
)

print("\nEvaluation Results")

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
